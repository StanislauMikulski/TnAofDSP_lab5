```python
fs = 1000 
N = 1024 
t = np.arange(N) / fs

signal = np.zeros(N)
signal[0] = 1.0 

signal0 = (1.0 / 1e-5) * np.exp(-np.pi * (t / 1e-5) ** 2)

fft_values = np.fft.rfft(signal)
frequencies = np.fft.rfftfreq(N, 1 / fs)

phases = np.angle(fft_values)
amplitudes = np.abs(fft_values)


signal0 = (1.0 / 1e-5) * np.exp(-np.pi * (t / 1e-5) ** 2)
```

```python
fig1, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 6))

ax1.plot(t, signal, color="crimson", linestyle="--", label="Дискретный импульс x(t)")
ax1.set_xlabel("Время t (с)")
ax1.set_ylabel("Амплитуда")
ax1.set_title("Исходный сигнал x(t)")
ax1.legend()

ax2.plot(t, signal0, color="forestgreen", linestyle="-.", label="Аналитическая запись")
ax2.set_xlabel("Время t (с)")  
ax2.set_ylabel("Значение функции")
ax2.set_title("Аналитическая аппроксимация дельта-функции")
ax2.legend()    

plt.tight_layout()
save_and_show(fig1, name="2_1_1")


fig2, (ax3, ax4) = plt.subplots(2, 1, figsize=(9, 6))

ax3.plot(frequencies, phases, color="crimson", linestyle="--", label="Фазовая характеристика")
ax3.set_xlabel("Частота (Hz)")
ax3.set_ylabel("Фаза")
ax3.set_title("Фаза сигнала")
ax3.legend()

ax4.plot(frequencies, amplitudes, color="forestgreen", linestyle="-.", label="Амплитудный спектр")
ax4.set_xlabel("Частота (Hz)")
ax4.set_ylabel("Амплитуда |X(f)|")
ax4.set_title("Амплитуда сигнала")
ax4.legend()

plt.tight_layout()
save_and_show(fig2, name="2_1_2")
```
