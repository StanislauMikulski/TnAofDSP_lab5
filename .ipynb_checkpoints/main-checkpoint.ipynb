## 1.1 Предсказание спектра

Глядя только на график своего сигнала $x(t) = (2 - 2t)u(t - 0{,}5) + u(0{,}5 - t)$, ответьте:

1) **Чему равна постоянная составляющая $a_0 / 2$:**
   Величина $a_0 / 2$ равна среднему значению сигнала за период, то есть площади под графиком, делённой на $T_0 = 1\text{ с}$. Площадь складывается из прямоугольника и треугольника:
   * прямоугольник: $S_1 = 0{,}5 \cdot 1 = 0{,}5$;
   * треугольник: $S_2 = 0{,}5 \cdot 0{,}5 \cdot 1 = 0{,}25$.
   $$\frac{a_0}{2} = 0{,}75 \implies a_0 = 1{,}5$$

2) **Будут ли нулевыми все $a_k$ или все $b_k$, и по какой симметрии:**
   Сигнал не является чётным: график не зеркален относительно оси $t = 0$, то есть $x(-t) \neq x(t)$. Сигнал не является и нечётным: $x(-t) \neq -x(t)$, к тому же у нечётного сигнала среднее значение равно нулю, а здесь $a_0/2 = 0{,}75 \neq 0$. Следовательно, ненулевыми будут и $a_k$, и $b_k$.

3) **Есть ли у периодического продолжения разрыв или излом и, следовательно, как будут убывать коэффициенты — как $1/k$ или как $1/k^2$:**
   При периодическом продолжении сигнала возникает скачок от $0$ до $1$: в конце периода значение равно $0$, а в начале следующего — $1$. В точке $t = 0{,}5$ оба куска сигнала равны $1$, то есть там сигнал непрерывен и есть только излом. Для разрывного сигнала коэффициенты Фурье убывают как $1/k$.

## 1.2 Аналитический расчёт

Сигнал задан выражением:
$$x(t) = (2 - 2t)u(t - 0{,}5) + u(0{,}5 - t)$$

Единичные скачки здесь работают как переключатели: при $t < 0{,}5$ слагаемое $u(t - 0{,}5)$ равно нулю, а $u(0{,}5 - t)$ равно единице, при $t > 0{,}5$ — наоборот. Поэтому на периоде сигнал состоит из двух кусков:
$$x(t) = \begin{cases} 1, & 0 \le t < 0{,}5 \\ 2 - 2t, & 0{,}5 \le t < 1 \end{cases}$$

Период $T_0 = 1\text{ с}$, поэтому $\omega_k = \frac{2\pi k}{T_0} = 2\pi k$. Формулы принимают вид:
$$a_k = 2 \int_{0}^{1} x(t)\cos(2\pi kt)\,dt$$
$$b_k = 2 \int_{0}^{1} x(t)\sin(2\pi kt)\,dt$$

### Постоянная составляющая $a_0$
$$a_0 = 2 \int_{0}^{0{,}5} 1 \cdot dt + 2 \int_{0{,}5}^{1} (2 - 2t)\,dt$$

Первый интеграл:
$$\int_{0}^{0{,}5} dt = \Big[t\Big]_{0}^{0{,}5} = 0{,}5 - 0 = 0{,}5$$

Второй интеграл:
$$\int_{0{,}5}^{1} (2 - 2t)\,dt = \Big[2t - t^2\Big]_{0{,}5}^{1} = (2 \cdot 1 - 1^2) - (2 \cdot 0{,}5 - 0{,}5^2) = 1 - 0{,}75 = 0{,}25$$

Складываем:
$$a_0 = 2(0{,}5 + 0{,}25) = 2 \cdot 0{,}75 = 1{,}5 \implies \frac{a_0}{2} = 0{,}75$$
Это совпадает со значением, предсказанным в п. 1.1 по площади под графиком.

*(Справочно: $\sin(\pi k) = 0$, $\sin(2\pi k) = 0$, $\cos(2\pi k) = 1$, $\cos(\pi k) = (-1)^k$)*

### Коэффициенты $a_k$
$$a_k = 2 \int_{0}^{0{,}5} 1 \cdot \cos(2\pi kt)\,dt + 2 \int_{0{,}5}^{1} (2 - 2t)\cos(2\pi kt)\,dt = 2(I_1 + I_2)$$

Интеграл $I_1$:
$$I_1 = \left[ \frac{\sin(2\pi kt)}{2\pi k} \right]_{0}^{0{,}5} = \frac{\sin(2\pi k \cdot 0{,}5) - \sin(0)}{2\pi k} = \frac{\sin(\pi k) - 0}{2\pi k} = 0$$

Интеграл $I_2$ интегрируем по частям ($\int u\,dv = uv - \int v\,du$):
* $u = 2 - 2t \implies du = -2\,dt$
* $dv = \cos(2\pi kt)\,dt \implies v = \frac{\sin(2\pi kt)}{2\pi k}$

$$I_2 = \left[ (2 - 2t)\frac{\sin(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} - \int_{0{,}5}^{1} \frac{\sin(2\pi kt)}{2\pi k} \cdot (-2)\,dt$$
$$I_2 = \left[ (2 - 2t)\frac{\sin(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} + \frac{1}{\pi k} \int_{0{,}5}^{1} \sin(2\pi kt)\,dt$$

Подстановка поднеинтегральной границы:
$$\left[ (2 - 2t)\frac{\sin(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} = \frac{(2 - 2 \cdot 1)\sin(2\pi k)}{2\pi k} - \frac{(2 - 2 \cdot 0{,}5)\sin(\pi k)}{2\pi k} = 0 - 0 = 0$$

Оставшийся интеграл от синуса:
$$\int_{0{,}5}^{1} \sin(2\pi kt)\,dt = \left[ -\frac{\cos(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} = -\frac{\cos(2\pi k)}{2\pi k} - \left( -\frac{\cos(\pi k)}{2\pi k} \right) = \frac{(-1)^k - 1}{2\pi k}$$

Значит:
$$I_2 = 0 + \frac{1}{\pi k} \cdot \frac{(-1)^k - 1}{2\pi k} = \frac{(-1)^k - 1}{2\pi^2 k^2}$$

Собираем $a_k$ (так как $I_1 = 0$):
$$a_k = 2 \left( 0 + \frac{(-1)^k - 1}{2\pi^2 k^2} \right) = \frac{(-1)^k - 1}{\pi^2 k^2}$$

При чётном $k$ имеем $(-1)^k = 1$ и числитель равен нулю. При нечётном $k$ имеем $(-1)^k = -1$ и числитель равен $-2$:
$$a_k = \begin{cases} -\frac{2}{\pi^2 k^2}, & k \text{ нечётное} \\ 0, & k \text{ чётное} \end{cases}$$

### Коэффициенты $b_k$
Разбиваем интеграл на два:
$$b_k = 2 \int_{0}^{0{,}5} 1 \cdot \sin(2\pi kt)\,dt + 2 \int_{0{,}5}^{1} (2 - 2t)\sin(2\pi kt)\,dt = 2(J_1 + J_2)$$

Интеграл $J_1$:
$$J_1 = \left[ -\frac{\cos(2\pi kt)}{2\pi k} \right]_{0}^{0{,}5} = -\frac{\cos(\pi k)}{2\pi k} - \left( -\frac{\cos(0)}{2\pi k} \right) = \frac{1 - (-1)^k}{2\pi k}$$

Интеграл $J_2$:
* $u = 2 - 2t \implies du = -2\,dt$
* $dv = \sin(2\pi kt)\,dt \implies v = -\frac{\cos(2\pi kt)}{2\pi k}$

$$J_2 = \left[ -(2 - 2t)\frac{\cos(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} - \int_{0{,}5}^{1} \left( -\frac{\cos(2\pi kt)}{2\pi k} \right) \cdot (-2)\,dt$$
$$J_2 = \left[ -(2 - 2t)\frac{\cos(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} - \frac{1}{\pi k} \int_{0{,}5}^{1} \cos(2\pi kt)\,dt$$

Упростим подстановку:
$$\left[ -(2 - 2t)\frac{\cos(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} = -\frac{(2 - 2 \cdot 1)\cos(2\pi k)}{2\pi k} - \left( -\frac{(2 - 2 \cdot 0{,}5)\cos(\pi k)}{2\pi k} \right) = 0 + \frac{1 \cdot (-1)^k}{2\pi k} = \frac{(-1)^k}{2\pi k}$$

Интеграл от косинуса:
$$\int_{0{,}5}^{1} \cos(2\pi kt)\,dt = \left[ \frac{\sin(2\pi kt)}{2\pi k} \right]_{0{,}5}^{1} = \frac{\sin(2\pi k) - \sin(\pi k)}{2\pi k} = 0$$

Следовательно, $J_2 = \frac{(-1)^k}{2\pi k}$. Складываем:
$$b_k = 2 \left( \frac{1 - (-1)^k}{2\pi k} + \frac{(-1)^k}{2\pi k} \right) = 2 \cdot \frac{1 - (-1)^k + (-1)^k}{2\pi k} = 2 \cdot \frac{1}{2\pi k} = \frac{1}{\pi k}$$

---

### Результат

* $a_0 = 1{,}5$
* $a_k = \frac{(-1)^k - 1}{\pi^2 k^2}$
* $b_k = \frac{1}{\pi k}$

Ряд Фурье имеет вид:
$$x(t) = \frac{3}{4} - \frac{2}{\pi^2} \sum_{k=1,3,5,\dots} \frac{\cos(2\pi kt)}{k^2} + \frac{1}{\pi} \sum_{k=1}^{\infty} \frac{\sin(2\pi kt)}{k}$$

**Значения первых коэффициентов:**
* $a_1 \approx -0{,}2026$, $b_1 \approx 0{,}3183$;
* $a_2 = 0$, $b_2 \approx 0{,}1592$;
* $a_3 \approx -0{,}0225$, $b_3 \approx 0{,}1061$;
* $a_4 = 0$, $b_4 \approx 0{,}0796$;
* $a_5 \approx -0{,}0081$, $b_5 \approx 0{,}0637$.

**Сопоставление с п. 1.1:** 
Среднее значение $a_0/2 = 0{,}75$ совпало с предсказанным. Коэффициенты ненулевые, как и ожидалось для сигнала без чётности и нечётности. Коэффициент $a_k$ убывает как $1/k^2$, а $b_k$ — как $1/k$.

## 2.1. Коэффициенты. 
Расчитайте $a_k$ и $b_k$ для $k = 0, 1, ..., 20$ численным интегрированием и постройте их графики с помощью функции `stem`, подписав оси. Сверьте с аналитическим результатом задания 1.2 и с предсказанием задания 1.1.

```python
T0 = 1.0

def f(x):
    x = np.mod(x, 1) # Сохранение периода [0,1]
    
    y = (2 - 2 * x) * np.heaviside(x - 0.5, 0.5) + np.heaviside(0.5 - x, 0.5)
    return y


def coefs(k):
    omega_k = k * (2 * np.pi / T0)

    x = np.linspace(0,T0,100)
    
    f_cos = lambda x: f(x) * np.cos(omega_k * x)
    f_sin = lambda x: f(x) * np.sin(omega_k * x) 

    if k == 0:
        return (2 / T0) * spi.quad(f_cos, 0, T0, points=[0.5])[0], 0 #  points=[0.5] нужен для того, чтобы разбить интеграл на два различных и избежать разрыв первого рода
    
    a_k = (2 / T0) * spi.quad(f_cos, 0, T0, points=[0.5])[0] # Берем только первое значение, так как spi.quad возвращает значение интеграла и значение ошибки
    b_k = (2 / T0) * spi.quad(f_sin, 0, T0, points=[0.5])[0]

    return a_k, b_k
```

```python
K = np.arange(21)
A = np.zeros(21)
B = np.zeros(21)
for idx, k in enumerate(K):
    A[idx], B[idx] = coefs(k)


fig, (Acoefs, Bcoefs) = plt.subplots(2, 1, figsize=(9, 6))

Acoefs.stem(K, A, basefmt="C0-")
Acoefs.set_ylabel("A coefs")
Acoefs.set_title("Коэффициенты А")

Bcoefs.stem(K, B, basefmt="C1-", linefmt="C1-", markerfmt="C1o")
Bcoefs.set_xlabel("k")
Bcoefs.set_ylabel("B coefs")
Bcoefs.set_title("Коэффициенты B")

plt.tight_layout()
save_and_show(fig, name="2_1_coefs")
```

## 2.2. Частичные суммы.  
Постройте $S_n(t)$ для $n ∈ {1, 3, 10, 30}$ на интервале $t ∈ [−T_0; 2T_0]$.
Наблюдается ли эффект Гиббса? Если да — у каких точек сигнала и почему именно у них

```python
def Sn(t, n):
    K = np.arange(n)
    A = np.zeros(n)
    B = np.zeros(n)
    for idx, k in enumerate(K):
        A[idx], B[idx] = coefs(k)
        
    y = A[0] / 2
    for i in range(1, len(A)):
        omega_k = i * (2 * np.pi / T0)
        y += A[i] * np.cos(omega_k * t) + B[i] * np.sin(omega_k * t)
    return y

n = 250
t = np.linspace(-T0, 2*T0, n)
Sn1 = np.zeros(n)
Sn3 = np.zeros(n)
Sn10 = np.zeros(n)
Sn30 = np.zeros(n)
Sn8 = np.zeros(n)
F = np.zeros(n)

with ProcessPoolExecutor(max_workers=6) as executor: #Добавим многопоточность для выполнения паралелльных операций (последовательно ждать оч долго)
    for idx, k in enumerate(t):
        f0 = executor.submit(f, k) 
        f1 = executor.submit(Sn, k, 1)
        f3 = executor.submit(Sn, k, 3)
        f10 = executor.submit(Sn, k, 10)
        f30 = executor.submit(Sn, k, 30)
        f8 = executor.submit(Sn, k, 8)
        
        Sn1[idx] = f1.result()
        Sn3[idx] = f3.result()
        Sn10[idx] = f10.result()
        Sn30[idx] = f30.result()
        Sn8[idx] = f8.result()
        F[idx] = f0.result()
```

```python
fig, ax = plt.subplots(figsize=(9, 6))    
plt.plot(t, F, color="dodgerblue", linewidth=2.5, label="Функция")
plt.plot(t, Sn1, color="crimson", linestyle="--", label="1 Гармоника")
plt.plot(t, Sn3, color="forestgreen", linestyle="-.", label="3 Гармоники")
plt.plot(t, Sn10, color="darkorange", linestyle=":", label="10 Гармоник")
plt.plot(t, Sn30, color="black", linestyle="-", alpha=0.8, label="30 Гармоник")

plt.xlabel("$t$")
plt.ylabel("$S_n(t)$")
plt.title("Проверка эффекта Гиббса")
plt.legend()
plt.tight_layout()
save_and_show(fig, name="2_2_Gibbs")
```

```python
f_mean = F.mean()
print(f"{f_mean:<12.4f} - Среднее значение функции\n")

for n, sn in [(1, Sn1), (3, Sn3), (10, Sn10), (30, Sn30)]:
    sn_mean = sn.mean()
    diff = -((sn_mean / f_mean) - 1) 
    
    print(f"{sn_mean:<12.4f} - Среднее значение суммы из {n} гармоник")
    print(f"{diff:<12.2%} - Разница суммы {n} гармоники и функции\n")
```

## 2.4. Спектр. 
Рассчитайте $A_k$ и $\varphi_k$ по формуле (5) для $k = 0, 1, \dots, 20$, постройте амплитудный и фазовый спектры. Гармоника с какой частотой (в герцах) имеет наибольшую амплитуду? Совпадает ли она с основным тоном?

$$A_k = \sqrt{a_k^2 + b_k^2}, \quad \varphi_k = \operatorname{atan2}(-b_k, a_k)$$

```python
amplitude = np.zeros(len(A))
phase = np.zeros(len(A))

for i in range(1, len(A)):
    amplitude[i] = math.sqrt(A[i]**2 + B[i]**2)
    phase[i] = math.atan2(-B[i], A[i])


fig, (Amplitude_graph, Phase_graph) = plt.subplots(2, 1, figsize=(9, 6))

Amplitude_graph.stem(K[1:], amplitude[1:], basefmt="C0-")
Amplitude_graph.set_ylabel("amplitude")
Amplitude_graph.set_title("Амплитудный спектр")

Phase_graph.stem(K[1:], phase[1:], basefmt="C1-", linefmt="C1-", markerfmt="C1o")
Phase_graph.set_xlabel("k")
Phase_graph.set_ylabel("phase")
Phase_graph.set_title("Фазовый спектр")

plt.tight_layout()
save_and_show(fig, name="2_4_amplitude")
```

## 3.2. Слышен ли эффект Гиббса.
Сравните на слух меандр и его частичную сумму $S_{51}$, у
которой выброс составляет почти 9 %. Слышна ли разница? Объясните результат, вспомнив,
что энергия выброса стремится к нулю, хотя высота — нет.

```python
duration = 3
f_meander = 50
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

meander = sps.square(2 * np.pi * f_meander * t, duty=0.5) 

yf = np.fft.fft(meander)
xf = np.fft.fftfreq(len(t), 1 / fs)

max_freq = 51 * f_meander
yf_truncated = yf.copy()
yf_truncated[np.abs(xf) > max_freq] = 0

Sn51 = np.real(np.fft.ifft(yf_truncated))

save_wav(meander, fs, "meander.wav")
save_wav(Sn51, fs, "Sn51.wav")

print("Меандр:"); display(Audio(meander, rate=fs))
print("Sn51:"); display(Audio(Sn51, rate=fs))
```
